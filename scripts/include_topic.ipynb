```python
import pandas as pd 
```

```python
prompts = pd.read_csv('../data/prompts.csv')
template_statements = pd.read_excel('../data/templates.xlsx', sheet_name='Statements')
template_terms = pd.read_excel('../data/templates.xlsx', sheet_name='Terms')
template_topics = pd.read_excel('../data/templates.xlsx', sheet_name='Topics')
```

```python
def assign_topic_term(prompt):
    for i, row in template_terms.iterrows():
        if row.TERM in prompt and "Define" in prompt:
            return str(row.ID_specific_topic)
    
prompts['topic'] = prompts.prompt.apply(assign_topic_term)
```

```python
def assign_topic_statement(prompt):
    for i, row in template_statements.iterrows():
        if row['MAIN STATEMENT'] in prompt:
            topic = str(row.ID_specific_topic)
            if not pd.isna(row['ID_specific_topic_2']):
                topic += '-' +  str(row['ID_specific_topic_2'])
            return topic

prompts.loc[prompts.topic.isna(), 'topic'] = prompts[prompts.topic.isna()].prompt.apply(assign_topic_statement)
```

```python
prompts.to_csv('../data/prompts_topic.csv', index = None)
```
